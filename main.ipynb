sorting by Category descending and UnitPrice_BDT ascending.

```python
df.sort_values(['Category', 'UnitPrice_BDT'], ascending=[0,1], inplace=True)
df.head()
```

### filtering data

```python
delivered = df.loc[df['Order Status'] == 'Delivered']
delivered.iloc[:,[3,4,5,8,10]]
```

```python
cancelled = df.loc[df['Order Status'] == 'Cancelled']
cancelled.iloc[:,[3,4,5,8,10]]
```

```python
# unit price in at least 10000

at_least_10_k = df.loc[df['UnitPrice_BDT'] >= 10000]
at_least_10_k.iloc[:,[5,8]]
```

```python
pending_or_cancelled = df.loc[(df['Order Status'] == 'Pending') | (df['Order Status'] == 'Cancelled')]
pending_or_cancelled.count()
```
